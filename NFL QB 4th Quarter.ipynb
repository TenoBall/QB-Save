```python
import nflreadpy as nfl

import polars as pl

# 1. Fetch 2025 play-by-play data (returns a Polars DataFrame)
pbp_2025 = nfl.load_pbp(seasons=[2025])

# 2. Filter for Rams offensive plays in the 4th quarter
rams_q4 = pbp_2025.filter(
    (pl.col("posteam") == "LA") & 
    (pl.col("qtr") == 4)
)

# 3. Select core columns
q4_plays = rams_q4.select([
    "game_id",
    "game_date",
    "quarter_seconds_remaining",
    "down",
    "ydstogo",
    "play_type",
    "desc",
    "yards_gained",
    "epa"
])

# Convert to pandas if you prefer working with Pandas DataFrames
df_pandas = q4_plays.to_pandas()
print(df_pandas.head())
```
